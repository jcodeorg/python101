## くりかえし
- くりかえして、何度も勝負できるようにしてみましょう


```python
print("くりかえしを開始します！")
while True:   # ずっと くりかえす
    print("0: グー, 1: チョキ, 2: パー")
    s = input("あなたの手を数字で入力してください: ")  # 入力を受け取る
    # もし入力が空文字（何も入力していない）なら、くりかえしを終了する
    if s == "":
        break
    # もし入力が 0, 1, 2 以外の値なら、くりかえしを続ける
    if s not in ["0", "1", "2"]:
        print("無効な入力です。もう一度入力してください。")
        continue
    player = int(s)     # 入力文字列を整数に変換
    print("入力された値は：", player)
print("くりかえしを終了します！")

```
