## キーボード操作（ショートカット）

文字を選択したり、切り貼りをしたりするときは、ショートカットを使うととても便利です。

多くのショートカットは、ほかのアプリでも共通して使えます。

| キー操作 | 動作 |
|:--:|:--|
| [shift]+[↑] | 行選択 |
| [shift]+[↓] | 行選択 |
| [shift]+[←] | 文字選択 |
| [shift]+[→] | 文字選択 |
| [ ctrl ]+[ a ] | 全選択 |
| [ ctrl ]+[ x ] | カット（切り取り） |
| [ ctrl ]+[ c ] | コピー |
| [ ctrl ]+[ v ] | ペースト（貼り付け） |
| [ ctrl ]+[ z ] | 取消（とりけし） |
| [ tab  ] | インデント（字下げ。スペース4つ） |
| [shift]+[ tab ] | インデント解除 |

※『＋』は「押したまま」と言う意味です。例えば [ ctrl ]+[ c ] は、 [ ctrl ]を押したまま（離さずに） [ c ] を押し、すぐに [ c ] を離します。

※ マックでは [ ctrl ] では無く ⌘ キーを使います。

## 7. 計算結果を表示する
`print(1 + 1)` は、1 + 1 の結果である 2 を表示します。

また、掛け算や割り算は、足し算や引き算よりも先に計算されます。
ただし、丸かっこを使うことで、計算の順番を変えることができます。

下のプログラムの数字を変更して、すべて計算結果として 9 が表示されるようにしてみましょう。

````python
print(5 + 2)                # 7  -> 9  足し算
print(5 - 2)                # 3  -> 9  引き算
print(5 * 2)                # 10 -> 9  掛け算（ * は「アスタリスク」）
print(5 / 2)                # 2.5-> 9  割り算
print(5 // 2)               # 2  -> 9  割り算をして、小数点以下を切り捨てた値（あたい）
print(5 % 2)                # 1  -> 9  割り算のあまり
print(5 ** 2)               # 25 -> 9  5の2乗（累乗・べき乗）
print( (5 + 4 + 6) / 3 )    # 5  -> 9  5, 4, 6 の平均
````

## 9. 文字列の演算

文字列は、以下のような演算ができます。
- +（文字列の結合）
- *（文字列の繰り返し）

下のプログラムを参考にして、文字列の結合と繰り返しを使い、次の文字列を表示してみよう。

*   ドキドキワクワク
*   ふわふわの雲
*   カリカリサクサクのスナック

````python
print("Hello" + "World")            # 文字列の結合
print("Hi!" * 3)                    # 文字列の繰り返し
print("ドキ" * 2 + "ワク" * 2)      # 文字列の繰り返しと結合
````
