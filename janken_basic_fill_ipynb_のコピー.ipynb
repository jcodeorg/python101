# じゃんけんゲーム（基本版・穴埋め）
このノートブックでは、Python を使ってじゃんけんゲームの基本版を作成します。
下線の部分 ____ を埋めて、じゃんけんゲームを完成させましょう。

## 今日やること
- リストを使って手を管理する
- random を使ってコンピュータの手を決める
- 条件分岐で勝敗判定を行う
- 入力と出力を使ってゲームを動かす

---
## ① ライブラリの読み込み
まずは random を読み込みます。

```python
import random  # random を読み込みます
```

## ② じゃんけんの手をリストで作る
リスト hands に「グー」「チョキ」「パー」を入れます。
文字列はダブルクォーテーションで囲んで入力することに注意してください。

```python
hands = ["グー", "チョキ", "パー"]  # グー・チョキ・パーを入れる
```

## ③ プレイヤーの入力を受け取る
input() を使って、数字（0, 1, 2）を入力してもらいます。

```python
print("0: グー, 1: チョキ, 2: パー")
player = int(input("あなたの手を数字で入力してください: "))  # 入力を受け取る
```

## ④ コンピュータの手をランダムに決める
random.randint(0, 2) を使います。

```python
computer = random.randint(0, 2)  # コンピュータの手を決める
```

## ⑤ 手を表示する
リスト hands を使って、グー・チョキ・パーのいずれかを表示します。

```python
print("あなた：", hands[player])
print("コンピュータ：", hands[computer])
```

## ⑥ 勝敗判定（条件分岐）
ここがじゃんけんのメイン部分です。
まずは「あいこ」の条件を書きます。

```python
if player == computer:  # あなたとコンピュータが同じ手の場合
    print("あいこです")
elif player == 0 and computer == 1: # あなたがグーでコンピュータがチョキの場合
    print("あなたの勝ち！")
elif player == 1 and computer == 2: # あなたがチョキでコンピュータがパーの場合
    print("あなたの勝ち！")
elif player == 2 and computer == 0: # あなたがパーでコンピュータがグーの場合
    print("あなたの勝ち！")
else:
    print("あなたの負け…")  # それ以外の場合は？
```

---
## ここまでで基本版が完成！
下のセルに、いままでのコードを全てコピペして、じゃんけんゲームを動かしてみましょう。

```python
# じゃんけんゲーム基本版
import random  # random を読み込みます
hands = ["グー", "チョキ", "パー"]  # グー・チョキ・パーを入れる
while True:
    print("0: グー, 1: チョキ, 2: パー")
    player = int(input("あなたの手を数字で入力してください: "))  # 入力を受け取る
    computer = random.randint(0, 2)  # コンピュータの手を決める
    print("あなた：", hands[player])
    print("コンピュータ：", hands[computer])
    if player == computer:  # あなたとコンピュータが同じ手の場合
        print("あいこです")
    elif player == 0 and computer == 1: # あなたがグーでコンピュータがチョキの場合
        print("あなたの勝ち！")
    elif player == 1 and computer == 2: # あなたがチョキでコンピュータがパーの場合
        print("あなたの勝ち！")
    elif player == 2 and computer == 0: # あなたがパーでコンピュータがグーの場合
        print("あなたの勝ち！")
    else:
        print("あなたの負け…")  # それ以外の場合は？
```
